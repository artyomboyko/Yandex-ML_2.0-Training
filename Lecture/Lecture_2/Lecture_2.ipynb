До 2012 года в основном использовали рекуррентные нейронные сети (Recurrent neural network - RNN). Эти модели могли генерировать тексты в стиле Шекспира, LaTeX и даже код на C. Эти модели в основном на каждом шаге предсказывали следующее слово для текущей последовательности. Более формально, эти модели предсказыыавли следующий токен при условии текущей последовательности.

## 2. Обработка последовательностей

До текущего момента мы пытались найти оптимальное отображение из пространства объектов $X$ в пространство ответов $Y$. Какими-то линейными или нелинейными методами. Но главное, все наши данные были представленны в виде вектора в каком-то векторном пространстве. Каждый вектор представлял из себя набор признаков.

Тест в свою очередь это последовательность.

Последовательность (формально) - это набор из дискретного (конечного) множества элементов для которого задано отношение порядка. Слайд [15:39]

Последовательности могут быть переменной длинны. 

Если же мы говорим про текст, то это набор дискретных элементов из конечного множества (ограниченное количество слов) элементов. В тесте все слова пренадлежат словарю, словарь имеет конечный размер. Все слова независимы (до построения эмбеддингов).

Видео тоже конечная последовательность картинок. Действия пользователя в браузере - тоже последовательность. Действия автономных агентов тоже последовательность. Последовательности везде. 

Задача - научится обрабатывать последовательности переменной длинны используя уже известные методы. Мы уже имеем отображать последовательность слов в векторы (на прошлом занятии научились). То есть мы умеем преобразовывать слова в эмбеддинги.

Но перед этим разберем еще один важный момент - Марковское свойство

### Марковское свойство

Слайд [32:08]

В последовательности наблюдается Марковское свойство (или последовательность обладает марковским свойством), если для предсказания ее следующего состояния важно только конечное множество предыдущих состояний. То есть важна только фиксированная часть предыдущей истории.

В тесте марковское свойство не наблюдается, так как каждое следующее слово может опираться на любой элемент из предыдущего контекста (то есть количество предыдущих состояний неограничено). 

Важный момент, марковское свойство так же может быть если есть какая-то константа $k$ которой можно ограничить количество предыдущих состояний, от которых зависит следующее состояние.

### Как работать с последовательностями переменной длины

Мы могли бы обрабатывать последовательности, если бы они все были одинаковой длинны. Мы бы могли собрать их в вектор. Но текст - последовательность переменной длинны. Это проблема потому что неизвестно какой длинны будет итоговый вектор, следовательно размерность исходного пространства не фиксированна. Для решения этой проблемы есть два варианта:

Слайд [41:27]

1. Глобальный статистики - взять что-то, что можно посчитать по всей длинне последовательности (например среднее). Усреднить можно любое количество элементов. На выходе будет вектор фиксированного размера. Глобальные статистики.

2. Пошаговая обработка, так как последовательность дискретна (шаг за шагом), то мы можем моделировать ее пошагово.

Предположим что мы умее кодировать текст в виде вектора. Это будет вектор $h$. К нам пришло новое слово $x_t$ и мы хотим обогатить вектор контекста модели с помощью этого нового слова. Если бы мы смогли каким-то образом задать функцию которая берет предыдущий вектор контекста и новый вектор слова и порождает новый вектор контекста, то это было бы отлично!

### Затухающий градиент

Эта проблема возникает из за функции активации, это как правило либо сигмоида, либо гиперболический тангенс. Чем глубже нейронная сеть, тем больше частных производных необходимо посчитать. Чем большо частных производных необходимо посчитать, тем меньше будет градиент из за функции активации.

Затухание градиента это проблема которая может привести к тому, что начальные слои нейронной сети будут мало обновляться (градиент доходящий до них будет маленький). То есть они не обучаться и будут из структурированных данных делать мусор.
