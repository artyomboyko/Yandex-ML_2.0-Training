## 2. Лучевой поиск (beam search)

Суть в том, что если мы ошибёмся на одном из шагов, то на следующем шаге правдоподобие этой последовательности сисильно упадет. А у правильного "луча" правдоподобие не упадет.

Лучевой поиск позволяет оценить несколько возможных последовательносте и выбрать из них наиболее правдоподобную. При этом мерилом качества выступает не внешняя метрика. Сама модель оценивает вероятность каждого из токенов и может посчитать совместную вероятность всей последовательности. 

У лучевого поиска есть один недостаток, этот подход предпочитает короткие последовательности длинным. Так как сумма логарифмов вероятностей (совместная вероятность) длинных последовательностей будет меньш чем у коротких последовательностей.
Эту проблему в лучевом поиске можно решить по разному, обычно с помощью нормализации совместной вероятности последовательности на ее длину (делим правдоподобие последовательности на её длинну).
