# Проверка Pytorch

```python
import torch

cuda_available = 'available' if torch.cuda.is_available() else 'not available'
cuda_device_count = torch.cuda.device_count()
cuda_current_device_name = torch.cuda.get_device_name(torch.cuda.current_device())

print(f'CUDA {cuda_available}, CUDA devices found: {cuda_device_count}, CUDA current device: {cuda_current_device_name}')
```
