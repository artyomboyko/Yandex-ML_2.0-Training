## Правдоподобие / Likelihood [00:10:00]

Пусть задана выборка X, Y и модель с параметрами $\theta$.

Правдоподобием будем называть 

$L(\theta | X, Y) = P(X, Y | \theta) \rightarrow \max_{\theta}$

помним про i.i.d.

$P(X, Y | \theta) = \prod_{i} P(x_i, y_i | \theta)$

$\log L(\theta | X, Y) = \log P(X, Y | \theta) = \sum_{i} \log P(x_i, y_i | \theta)$
