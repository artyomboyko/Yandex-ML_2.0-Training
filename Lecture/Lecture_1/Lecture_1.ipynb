### I.I.D: Независимые и Одинаково Распределенные Случайные Величины

**IID** (англ. Independent and Identically Distributed random variables) расшифровывается как **Независимые и Одинаково распределенные случайные величины** 

Это ключевое понятие в теории вероятностей и статистике, описывающее набор случайных величин с двумя важными свойствами:
**1. Независимость:** 
    * Значение любой из случайных величин не влияет на значения остальных. 
    * Знание значения одной величины не дает информации о значениях других величин в наборе.
**2. Одинаковая распределенность:** 
    * Все случайные величины имеют одинаковое распределение вероятностей. 
    * Вероятность того, что любая из величин примет определенное значение, одинакова для всех величин.

Пример:
Представьте, что вы подбрасываете монетку 10 раз. Результаты каждого броска (орел или решка) - это набор из 10 i.i.d. случайных величин. 
* **Независимость:** Результат каждого броска не зависит от других бросков.
* **Одинаковая распределенность:** Вероятность орла или решки одинакова для каждого броска (если монета честная).

**Зачем нужно свойство i.i.d.?**    
Свойство i.i.d. упрощает математический анализ и моделирование случайных процессов. Многие статистические методы и теоремы основаны на предположении i.i.d. данных. 

**Примеры использования i.i.d. в статистике:**    
* **Центральная предельная теорема:** сумма большого числа i.i.d. случайных величин имеет приблизительно нормальное распределение.
* **Закон больших чисел:** среднее арифметическое большого числа i.i.d. случайных величин стремится к их математическому ожиданию.
* **Статистическое оценивание:** методы оценивания параметров (например, метод максимального правдоподобия) часто основаны на предположении i.i.d. выборки.

**Важно помнить:**   
В реальных задачах данные не всегда i.i.d. Например, временные ряды часто имеют корреляцию между соседними значениями, нарушая независимость. 
В таких случаях нужны более сложные модели и методы анализа, учитывающие зависимость между данными.

### Правдоподобие

Пусть задана выборка X, Y и модель с параметрами $\theta$.

Правдоподобием будем называть 

$L(\theta | X, Y) = P(X, Y | \theta) \rightarrow \max_{\theta}$

помним про i.i.d.

$P(X, Y | \theta) = \prod_{i} P(x_i, y_i | \theta)$

$\log L(\theta | X, Y) = \log P(X, Y | \theta) = \sum_{i} \log P(x_i, y_i | \theta)$
