## Загрука данных

Загрузим датасет:

```python
train_df = pd.read_csv('./ml_trains_data/train.tsv', sep = '\t', encoding='utf-8')
test_df = pd.read_csv('./ml_trains_data/eval.tsv', sep = '\t', encoding='utf-8')
```

```python
train_df.rename(columns={'YTID': 'file'}, inplace=True)

train_df['file'] = 'ml_trains_data/clips/' + train_df['file'] + '.flac'
train_df.to_csv('./ml_trains_data/clips/train_metadata.csv', sep = ',', index = False)

test_df.rename(columns={'YTID': 'file'}, inplace=True)
test_df['label'] = None
test_df['file'] = 'ml_trains_data/clips/' + test_df['file'] + '.flac'
test_df.to_csv('./ml_trains_data/clips/test_metadata.csv', sep = ',', index = False)
```
