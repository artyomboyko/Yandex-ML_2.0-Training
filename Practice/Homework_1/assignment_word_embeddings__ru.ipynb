#### Подзадача №1: subsampling
Для того, чтобы сгладить разницу в частоте встречаемости слов, необходимо реализовать механизм subsampling'а.
Для этого вам необходимо реализовать функцию ниже.

Вероятность **исключить** слово из обучения (на фиксированном шаге) вычисляется как
$$
P_\text{drop}(w_i)=1 - \sqrt{\frac{t}{f(w_i)}},
$$
где $f(w_i)$ – нормированная частота встречаемости слова, а $t$ – заданный порог (threshold).

## Решение

````python
def subsample_frequent_words(word_count_dict, threshold = 1e-5):  # threshold=8e-3
    """
    Вычисление вероятностей подвыборки для слов на основе их частот.

    Эта функция используется для определения вероятности сохранения слова в датасете при подвыборке частых слов.
    Используемый метод вдохновлен подходом к подвыборке в Word2Vec, где частота каждого слова влияет на вероятность
    его сохранения.

    Параметры:
    - word_count_dict (dict): Словарь, в котором ключами являются слова, а значениями - счетчики этих слов.
    - threshold (float, необязательно): Пороговый параметр, используемый для настройки частоты поддискретизации слов.
                                   По умолчанию равен 1e-5.

    Возвращает:
    - dict: Словарь, в котором ключами являются слова, а значениями - вероятности сохранения каждого слова.

    Example:
    >>> word_counts = {'the': 5000, 'is': 1000, 'apple': 50}
    >>> subsample_frequent_words(word_counts)
    {'the': 0.028, 'is': 0.223, 'apple': 1.0}
    """
    total_words = sum(word_count_dict.values())
    keep_prob_dict = {}

    for word, count in word_count_dict.items():
        frequency = count / total_words
        # word_prob = ((count / (threshold * total_words))**0.5 + 1) * (threshold * total_words) / count
        # word_prob = 1 - (threshold / frequency)**0.5
        word_prob = (threshold / frequency)**0.5
        keep_prob_dict[word] = word_prob

    return keep_prob_dict
````

````python
word_counts = {'the': 5000, 'is': 1000, 'apple': 50}
subsample_frequent_words(word_counts, threshold=80e-4)
````

#### Подзадача №2: negative sampling
Для более эффективного обучения необходимо не только предсказывать высокие вероятности для слов из контекста, но и предсказывать низкие для слов, не встреченных в контексте. Для этого вам необходимо вычислить вероятност использовать слово в качестве negative sample, реализовав функцию ниже.

В оригинальной статье предлагается оценивать вероятности слов выступающих в качестве negative sample согласно распределению $P_n(w)$
$$
P_n(w) = \frac{U(w)^{3/4}}{Z},
$$

где $U(w)$ распределение слов по частоте (или, как его еще называют, по униграммам), а $Z$ – нормировочная константа, чтобы общая мера была равна $1$.

````python
def get_negative_sampling_prob(word_count_dict):
    """
    Вычисляет негативные вероятности выборки для слов на основе их частот.

    Эта функция корректирует частоту каждого слова, возведенную в степень 0,75, которая обычно используется в алгоритмах 
    типа Word2Vec для смягчения влияния очень частых слов. Затем она нормализует эти скорректированные частоты, 
    чтобы их сумма равнялась 1, формируя распределение вероятности, используемое для отрицательной выборки.

    Параметры:
    - word_count_dict (dict): Словарь, в котором ключами являются слова, а значениями - количество этих слов.

    Возвращает:
    - dict: Словарь, в котором ключами являются слова, а значениями - вероятности выбора каждого слова для отрицательной выборки.

    Example:
    >>> word_counts = {'the': 5000, 'is': 1000, 'apple': 50}
    >>> get_negative_sampling_prob(word_counts)
    {'the': 0.298, 'is': 0.160, 'apple': 0.042}
    """
    
    total_count = sum(word_count_dict.values())
    unigram_probs = {word: count / total_count for word, count in word_count_dict.items()}
    
    # Вычисляем нормировочную константу Z
    Z = sum([prob ** 0.75 for prob in unigram_probs.values()])
    
    negative_sampling_prob_dict = {word: (prob ** 0.75) / Z for word, prob in unigram_probs.items()} 

    return negative_sampling_prob_dict
````

Если все прошло успешно, функция ниже поможет вам с генерацией подвыборок (батчей).

````python
def generate_batch_with_neg_samples(
    context_pairs,
    batch_size,
    keep_prob_array,
    word_to_index,
    num_negatives,
    negative_sampling_prob_array,
):
    batch = []
    neg_samples = []

    while len(batch) < batch_size:
        center, context = random.choice(context_pairs)
        if random.random() < keep_prob_array[center]:
            batch.append((center, context))
            neg_sample = np.random.choice(
                range(len(negative_sampling_prob_array)),
                size=num_negatives,
                p=negative_sampling_prob_array,
            )
            neg_samples.append(neg_sample)
    batch = np.array(batch)
    neg_samples = np.vstack(neg_samples)
    return batch, neg_samples
````

Напомним, что в случае negative sampling решается задача максимизации следующего функционала:

$$
\mathcal{L} = \log \sigma({\mathbf{v}'_{w_O}}^\top \mathbf{v}_{w_I}) + \sum_{i=1}^{k} \mathbb{E}_{w_i \sim P_n(w)} \left[ \log \sigma({-\mathbf{v}'_{w_i}}^\top \mathbf{v}_{w_I}) \right],
$$

где:
- $\mathbf{v}_{w_I}$ – вектор центрального слова $w_I$,
- $\mathbf{v}'_{w_O}$ – вектор слова из контекста $w_O$,
- $k$ – число negative samplesЮ,
- $P_n(w)$ – распределение negative samples, заданное выше,
- $\sigma$ – сигмоида.

Вариант 3 (с логсигмоидой):

````python
class SkipGramModelWithNegSampling(nn.Module):
    def __init__(self, vocab_size, embedding_dim):
        super().__init__()
        # self.center_embeddings = torch.nn.Embedding(vocab_size, embedding_dim) # Оригинал
        # self.context_embeddings = torch.nn.Embedding(vocab_size, embedding_dim) # Оригинал
        self.center_embeddings = torch.nn.Embedding(vocab_size, embedding_dim, sparse=True) # Попробовать sparse=True
        self.context_embeddings = torch.nn.Embedding(vocab_size, embedding_dim, sparse=True) # Попробовать sparse=True



    def forward(self, center_words, pos_context_words, neg_context_words):
        center_embeds = self.center_embeddings(center_words)  # batch size, embedding_dim
        pos_context_embeds = self.context_embeddings(pos_context_words)  # batch size, embedding dim

        neg_context_embeds = self.context_embeddings(neg_context_words)  # batch_size, num_negatives, embedding_dim

        pos_dot_products = torch.sum(center_embeds * pos_context_embeds, dim=-1)  # batch_size
        pos_scores = F.logsigmoid(pos_dot_products)  # Применяем логсигмоиду к pos_scores

        center_embeds_expanded = center_embeds.unsqueeze(1)  # batch_size, 1, embedding_dim

        neg_dot_products = torch.bmm(neg_context_embeds, center_embeds_expanded.transpose(1, 2)).squeeze(2)  # batch_size, num_negatives
        neg_scores = F.logsigmoid(neg_dot_products)  # Применяем логсигмоиду к neg_scores УБРАЛ - !!!!

        return pos_scores, neg_scores
````

````python
vocab_size = len(word_to_index)
embedding_dim = 32
num_negatives = 15

model = SkipGramModelWithNegSampling(vocab_size, embedding_dim).to(device)
# optimizer = optim.Adam(model.parameters(), lr=0.05) # Попробовать SparceAdam с установкой в слоях эмбеддингов sparse=True
optimizer = optim.SparseAdam(model.parameters(), lr=0.05)
lr_scheduler = ReduceLROnPlateau(optimizer, factor=0.5, patience=150)
criterion = nn.BCEWithLogitsLoss()
````

````python
params_counter = 0
for weights in model.parameters():
    params_counter += weights.shape.numel()
assert params_counter == len(word_to_index) * embedding_dim * 2
````

````python

def train_skipgram_with_neg_sampling(
    model,
    context_pairs,
    keep_prob_array,
    word_to_index,
    batch_size,
    num_negatives,
    negative_sampling_prob_array,
    steps,
    optimizer=optimizer,
    lr_scheduler=lr_scheduler,
    device=device,
):
    pos_labels = torch.ones(batch_size).to(device)
    neg_labels = torch.zeros(batch_size, num_negatives).to(device)
    loss_history = []
    for step in tqdma(range(steps)):
        batch, neg_samples = generate_batch_with_neg_samples(
            context_pairs,
            batch_size,
            keep_prob_array,
            word_to_index,
            num_negatives,
            negative_sampling_prob_array,
        )
        center_words = torch.tensor([pair[0] for pair in batch], dtype=torch.long).to(
            device
        )
        pos_context_words = torch.tensor(
            [pair[1] for pair in batch], dtype=torch.long
        ).to(device)
        neg_context_words = torch.tensor(neg_samples, dtype=torch.long).to(device)

        optimizer.zero_grad()
        pos_scores, neg_scores = model(
            center_words, pos_context_words, neg_context_words
        )

        loss_pos = criterion(pos_scores, pos_labels)
        loss_neg = criterion(neg_scores, neg_labels)

        loss = loss_pos + loss_neg
        loss.backward()
        optimizer.step()

        loss_history.append(loss.item())
        lr_scheduler.step(loss_history[-1])

        if step % 100 == 0:
            print(
                f"Step {step}, Loss: {np.mean(loss_history[-100:])}, learning rate: {lr_scheduler._last_lr}"
            )
````

````python
torch.manual_seed(777)

steps = 100    # steps = 2500
batch_size = 512 # batch_size = 512
train_skipgram_with_neg_sampling(
    model,
    context_pairs,
    keep_prob_array,
    word_to_index,
    batch_size,
    num_negatives,
    negative_sampling_prob_array,
    steps,
)
````

Наконец, взглянем на ближайшие по косинусной мере слова. Функция реализована ниже.

````python
def find_nearest(word, embedding_matrix, word_to_index=word_to_index, k=10):
    word_vector = get_word_vector(word, embedding_matrix)[None, :]
    dists = F.cosine_similarity(embedding_matrix, word_vector)
    index_sorted = torch.argsort(dists)
    top_k = index_sorted[-k:]
    return [(index_to_word[x], dists[x].item()) for x in top_k.numpy()]
````
